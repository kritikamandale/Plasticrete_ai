Explores master_dataset.csv: coverage, distributions, and outliers.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from modules.m1_prediction.utils.data_loader import load_all_sources, print_coverage_report
from modules.m1_prediction.utils.schema import TARGET_COLUMNS, INPUT_FEATURES

master = load_all_sources('data/raw')
print_coverage_report(master)
```

```python
# Target distributions
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, col in zip(axes.ravel(), TARGET_COLUMNS):
    data = master[col].dropna()
    if len(data) > 0:
        ax.hist(data, bins=30, edgecolor='white', color='steelblue', alpha=0.8)
        ax.set_title(f'{col}\nn={len(data)}')
        ax.set_xlabel(col.split('_')[-1])
axes.ravel()[-1].axis('off')
plt.tight_layout()
plt.show()
```

```python
# Source distribution
master.groupby('source_doi').size().sort_values(ascending=False).plot(
    kind='barh', figsize=(10, 5), title='Rows per source'
)
plt.tight_layout()
plt.show()
```
